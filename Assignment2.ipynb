### 1 Input & output

```python
# Que. Read the messy blog post from a text(messy_blog.txt file)
```

```python
with open("messy_blog.txt","r") as f:
 blog=f.read()
 print(blog)
```

```python
# print the first 100 charactors of the blog post using string slicing

print(blog[:100])
```

### Q 2: String Cleaning

```python
# Removes the unwanted symbol

text=blog
text = text.replace("gr8","great")
text = text.replace("btw","By the way")
text = text.replace("idk", "I dont know")
text = text.replace("lol","laughing out loud")
text = text.replace("###","")
text = text.replace("!!!","")
text = text.replace("???","")
text = text.replace("@@@","")
print(blog)
```

# 3.formatting

```python
# remove duplicate blanck lines
blog = text. replace("\n\n","\n")
print(blog)
```

```python
# capitalize the text
blog = text.capitalize()
blog = text.title()
print(blog)
```

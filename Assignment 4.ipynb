## 2] Dictionaries (Key–Value Pairs, Methods)

1. Create a dictionary to store mobile phone prices. Example format: { 'iPhone': 50000, 'Samsung': 40000, ... }
2. During the sale, the price of "iPhone" is reduced to ₹48,000.
   ○ Update the dictionary accordingly.
3. Add the price for the newly arrived phone "Pixel" to the dictionary.


```python
price = {     #dictionary   # price is a key value
    "iphone":50000,
    "Samsung":40000, 
    "Vivo":30000,
    "Oppo":20000,
    "OnePlus" : 35000,
    }
print("original prices:", price)

price.update({"Sale on iphone":48000}) 
print("with discount price:",price)

price["Pixel:"]=25000
print("New item added:",price.items()) 
```

## 3]  Sets (Unique Elements, Set Operations)

   3. Perform the following operations:
      ○ Find the unique customers who visited the store today (Union).
      ○ Find the customers who visited both morning and evening (Intersection).


```python
Morning_customer = {"rohan","seema","advait","suman","hrutvik","shreya","harshda","naitik"}
Evening_customer = {"Kalanjali","naitik","harshda","anamika","suman"}
print("Morning customer",Morning_customer)
print("Evening customer",Evening_customer)
print("Visited store today: ",Morning_customer.union(Evening_customer))
print("Two times visited customer : ",Morning_customer.intersection(Evening_customer))
print(type(Morning_customer))
```

## 4] Comprehensions (Flash Sale Discount)
    1. Apply a Flash Sale discount of 10% on all mobile prices.
      ○ Create a new dictionary using dictionary comprehension to store discounted prices.
    2. Create a list containing only the discounted prices using list comprehension.


```python
#Apply a flash sale discount of 10% on all mobile phones
print(".....Flash sale discount....")
price = {                   # original prices dictionary
   "iPhone": 50000,
   "Samsung": 40000,
   "Vivo": 30000,
   "Oppo": 20000,
   "OnePlus": 35000,
   "pixel": 25000,
}

discounted_price = {brand : price*0.9 for brand ,price in price.items()} # showing ammount reduces price by 10%
print("Discounted prices for all phones:",discounted_price)
#discounted_list = [value * 0.9 for value in price.values()]
#print(discounted_list)
```

```python
#Create a list containing only the discounted prices using list comprehension.
price = {                   # original prices dictionary
   "iPhone": 50000,
   "Samsung": 40000,
   "Vivo": 30000,
   "Oppo": 20000,                          
   "OnePlus": 35000,
   "pixel": 25000,
}
#discounted_list = [price*0.1 for price in price.values()]
#print("List of Discounted prices:", discounted_price)

discount_amount = {price: value * 0.1 for price, value in price.items()}
print(discount_amount)
```

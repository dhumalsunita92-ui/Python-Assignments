```python
### The smart "ATM" simulator create a test based ATM program
#Q 1 set hardcoaded PIN(e.g 1234) and a starting balance (e.g 10000)

# Smart ATM Simulator using if-else


pin = "1234"          # fixed PIN
ATM_balance = 10000   # starting balance

# Step 1: PIN verification
entered_pin = input("Enter your PIN: ")

if entered_pin == pin:
    print("PIN verified, welcome to Smart ATM")
else:
    print("Invalid PIN. Exiting...")
    exit()

# Step 2: Menu
while True:
    print("\n---- Select your choice ----")
    print("1. Check balance")
    print("2. Withdraw money")
    print("3. Deposit money")
    print("4. Exit")

    choice = input("Enter your request number (1-4): ")

    if choice == "1":
        print(f"Your balance is: {ATM_balance}")

    elif choice == "2":
        amount = int(input("Enter amount to withdraw: "))
        if amount <= ATM_balance:
            ATM_balance -= amount
            print(f"{amount} withdrawn successfully. New balance: {ATM_balance}")
        else:
            print("Insufficient balance!")

    elif choice == "3":
        amount = int(input("Enter amount to deposit: "))
        ATM_balance += amount
        print(f"{amount} deposited successfully. New balance: {ATM_balance}")

    elif choice == "4":
        print("Thank you for using Smart ATM. Goodbye!")
        break

    else:
        print("Invalid choice, please try again.")
```

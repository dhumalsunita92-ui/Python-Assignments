### 5] special task : diwli offer(dictionary comprehension)
   - sharma ji wants to give a discount
   - use dictionary comprehension to create a new dictionary called offer menu prices
   - print the final total

```python
# offer menu dictionary
menu = {
    "samosa": 20,
    "chai": 10,
    "coffee": 15,
    "vada pav": 20,
    "idli": 30
}

orders = ["vada pav", "coffee", "samosa"]

discount_rate = 0.10

# Dictionary Comprehension
offer_menu_price = {
    item: price - (price * discount_rate)
    for item, price in menu.items()
}

print("Offer Menu")

for item, price in offer_menu_price.items():
    print(f"{item}: Rs.{price}")

final_total = 0

for item in orders:
    if item in offer_menu_price:
        final_total += offer_menu_price[item]
    else:
        print(f"{item} not available")
print(f"\nTotal Offer Menu Price: Rs.{final_total}")
```
